```python
from abc import ABC, abstractmethod

class Invoice(ABC):

    def __init__(self, product_name, amount):
        self.product_name = product_name
        self.__amount = amount

    def get_amount(self):
        return self.__amount

    @abstractmethod
    def calculate_tax(self):
        pass

    def generate_invoice(self):
        tax = self.calculate_tax()
        total = self.get_amount() + tax

        print("\n========== INVOICE ==========")
        print("Product Name :", self.product_name)
        print("Product Price: ₹", self.get_amount())
        print("Tax Amount   : ₹", round(tax, 2))
        print("Total Amount : ₹", round(total, 2))
        print("=============================")

class IndiaInvoice(Invoice):

    def calculate_tax(self):
        gst = self.get_amount() * 0.18
        print("GST (18%)    : ₹", round(gst, 2))
        return gst
        
class USAInvoice(Invoice):

    def calculate_tax(self):
        vat = self.get_amount() * 0.10
        state_tax = self.get_amount() * 0.05

        print("VAT (10%)    : ₹", round(vat, 2))
        print("State Tax(5%): ₹", round(state_tax, 2))

        return vat + state_tax

while True:

    print("\n===== E-COMMERCE INVOICE SYSTEM =====")
    print("1. India Invoice (GST)")
    print("2. USA Invoice (VAT + State Tax)")
    print("3. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":

        product = input("Enter Product Name: ")
        amount = float(input("Enter Product Price: ₹"))

        invoice = IndiaInvoice(product, amount)
        invoice.generate_invoice()

    elif choice == "2":

        product = input("Enter Product Name: ")
        amount = float(input("Enter Product Price: ₹"))

        invoice = USAInvoice(product, amount)
        invoice.generate_invoice()

    elif choice == "3":

        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```

```python
from datetime import datetime
trip_logs = []
RATE_PER_KM = 15
def calculate_fare(distance, hour):

    fare = distance * RATE_PER_KM


    if (7 <= hour <= 9) or (17 <= hour <= 21):
        fare *= 1.5

    return fare


def add_trip():

    driver_name = input("Enter Driver Name: ")
    distance = float(input("Enter Distance Travelled (km): "))
    hour = int(input("Enter Trip Hour (0-23): "))

    fare = calculate_fare(distance, hour)

    trip = {
        "driver": driver_name,
        "distance": distance,
        "hour": hour,
        "fare": fare
    }

    trip_logs.append(trip)

    print(f"Trip Recorded Successfully!")
    print(f"Fare = ₹{fare:.2f}")

def view_trips():

    if not trip_logs:
        print("No trips available.")
        return

    print("\n===== TRIP LOGS =====")

    for trip in trip_logs:
        print("--------------------------")
        print("Driver :", trip["driver"])
        print("Distance :", trip["distance"], "km")
        print("Hour :", trip["hour"])
        print("Fare : ₹", round(trip["fare"], 2))


def generate_report():

    total_revenue = 0

    for trip in trip_logs:
        total_revenue += trip["fare"]

    filename = "daily_revenue_report.txt"

    with open(filename, "w") as file:

        file.write("DAILY RIDE REVENUE REPORT\n")
        file.write("=========================\n")
        file.write(f"Date: {datetime.now()}\n\n")

        for trip in trip_logs:

            file.write(
                f"Driver: {trip['driver']} | "
                f"Distance: {trip['distance']} km | "
                f"Fare: ₹{trip['fare']:.2f}\n"
            )

        file.write("\n")
        file.write(f"TOTAL REVENUE = ₹{total_revenue:.2f}")

    print("\nReport Generated Successfully!")
    print("File Name:", filename)
    print(f"Total Revenue: ₹{total_revenue:.2f}")

while True:

    print("\n===== RIDE SHARING SYSTEM =====")
    print("1. Add Driver Trip")
    print("2. View Trip Logs")
    print("3. Generate Daily Revenue Report")
    print("4. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_trip()

    elif choice == "2":
        view_trips()

    elif choice == "3":
        generate_report()

    elif choice == "4":
        print("Exiting System...")
        break
```

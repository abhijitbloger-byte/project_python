```python
inventory = {
    "P101": {"name": "Laptop", "quantity": 20},
    "P102": {"name": "Mouse", "quantity": 50},
    "P103": {"name": "Keyboard", "quantity": 30}
}


def view_inventory():
    print("\n===== INVENTORY DASHBOARD =====")
    if not inventory:
        print("Inventory is empty.")
        return

    print(f"{'Product ID':<12}{'Product Name':<20}{'Quantity':<10}")
    print("-" * 42)

    for pid, details in inventory.items():
        print(f"{pid:<12}{details['name']:<20}{details['quantity']:<10}")


def restock_product():
    print("\n===== INBOUND LOGISTICS: RESTOCK / ADD PRODUCT =====")

    product_id = input("Enter Product ID: ")

    if product_id in inventory:
        qty = int(input("Enter quantity to add: "))
        inventory[product_id]["quantity"] += qty
        print("Stock updated successfully.")
    else:
        name = input("Enter Product Name: ")
        qty = int(input("Enter Initial Quantity: "))

        inventory[product_id] = {
            "name": name,
            "quantity": qty
        }

        print("New product added successfully.")


def process_sale():
    print("\n===== OUTBOUND LOGISTICS: PROCESS ORDER SALE =====")

    product_id = input("Enter Product ID: ")

    if product_id not in inventory:
        print("Product not found.")
        return

    qty = int(input("Enter quantity sold: "))

    if qty > inventory[product_id]["quantity"]:
        print("Insufficient stock.")
    else:
        inventory[product_id]["quantity"] -= qty
        print("Sale processed successfully.")
        print(f"Remaining stock: {inventory[product_id]['quantity']}")


def main():
    while True:
        print("\n=========================================")
        print(" SUPPLY CHAIN & INVENTORY MANAGEMENT SYSTEM")
        print("=========================================")
        print("1. View Inventory Dashboard")
        print("2. Inbound Logistics: Restock/Add Product")
        print("3. Outbound Logistics: Process Order Sale")
        print("4. Exit System")

        choice = input("\nEnter your choice (1-4): ")

        if choice == "1":
            view_inventory()

        elif choice == "2":
            restock_product()

        elif choice == "3":
            process_sale()

        elif choice == "4":
            print("Exiting system... Goodbye!")
            break

        else:
            print("Invalid choice. Please select between 1 and 4.")


if __name__ == "__main__":
    main()
```
